```python
# SECCIÓN 1: IMPORTAR LIBRERÍAS

# Este apartado sirve para importar las herramientas que
# utilizaremos durante todo el proyecto.

# Pandas permite leer y manipular archivos CSV.
import pandas as pd

# StandardScaler sirve para normalizar los datos antes
# de aplicar el algoritmo K-Means.
from sklearn.preprocessing import StandardScaler

# KMeans es el algoritmo que clasificará a los clientes
# en diferentes grupos según sus características.
from sklearn.cluster import KMeans
```

```python
# SECCIÓN 2: CARGAR EL ARCHIVO

# Este apartado sirve para leer el archivo CSV que contiene
# la información de los clientes y guardarla en una variable
# llamada "df" (DataFrame), que es donde trabajaremos los datos.

import pandas as pd

df = pd.read_csv("dataset_clientes.csv")

# Muestra las primeras 5 filas para comprobar que el archivo
# se cargó correctamente.
df.head()
```

```python
#SECCIÓN 3: EXPLORACIÓN DE LOS DATOS

# Este apartado permite conocer cuántas filas y columnas tiene
# el archivo, el tipo de datos de cada columna y si existen
# valores vacíos.

df.info()

# Cuenta cuántos valores faltantes hay en cada columna.
df.isnull().sum()
```

```python
# SECCIÓN 4: LIMPIEZA DE DATOS

# Este apartado elimina las filas que tengan información
# incompleta para trabajar únicamente con datos válidos.

df = df.dropna()

# También elimina registros duplicados, si existen.
df = df.drop_duplicates()

# Muestra cuántos registros quedaron después de la limpieza.
print(df.shape)
```

```python
# SECCIÓN 5: SELECCIÓN DE VARIABLES

# Este apartado selecciona únicamente las columnas que
# serán utilizadas para realizar la clasificación.

X = df[[
    "Compras_Mensuales",
        "Gasto_Mensual_MXN",
            "Dias_Desde_Ultima_Compra"
            ]]
```

```python
# SECCIÓN 6: NORMALIZACIÓN

# Este apartado hace que todas las variables tengan la
# misma escala para que ninguna influya más que otra.

escalador = StandardScaler()

X_escalado = escalador.fit_transform(X)
```

```python
# SECCIÓN 7: CLASIFICACIÓN CON K-MEANS

# Este apartado crea un modelo de K-Means que dividirá
# los clientes en tres grupos según sus características.

modelo = KMeans(n_clusters=3, random_state=42, n_init=10)

# Se ejecuta el algoritmo para asignar cada cliente a un grupo.
clusters = modelo.fit_predict(X_escalado)
```

```python
# SECCIÓN 8: AGREGAR LOS CLUSTERS

# Este apartado agrega una nueva columna llamada Cluster,
# indicando el grupo al que pertenece cada cliente.

df["Cluster"] = clusters

# Mostrar las primeras filas con la clasificación.
df.head()
```

```python
# SECCIÓN 9: ANÁLISIS DE RESULTADOS

# Este apartado calcula el promedio de cada grupo para
# conocer las características de cada tipo de cliente.

print(df.groupby("Cluster").mean(numeric_only=True))
```
